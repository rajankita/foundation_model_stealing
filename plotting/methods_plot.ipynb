```python
import matplotlib.pyplot as plt
import csv
import pandas as pd
import seaborn as sns
sns.set()
```

```python
filename = '../results/methods_indoor.csv'
out_file = '../results_camera_ready/methods_plot.png'

df0 = pd.read_csv(filename, index_col='Victim arch')
df0 = df0.drop('ViT-L/16', axis=1)
df0_t = df0.T
print(df0_t)
```

```python
x = df0.columns
print(x)
sns.set_style("whitegrid")
sns.set_context("paper", font_scale=2.2)

sns.set_palette("tab10", 4)
fig = plt.figure(figsize=(10,6))
ax = fig.gca()
l1 = ax.plot(x, df0.loc["Victim Acc"], lw=3, ls='--', marker='o', markersize='10', label='Victim Accuracy')
l2 = ax.plot(x, df0.loc["Random"], lw=3, ls='-', marker='v', markersize='10', label='Random')
l2 = ax.plot(x, df0.loc["k-Center"], lw=3, ls='-', marker='v', markersize='10', label='k-Center')
l4 = ax.plot(x, df0.loc["Entropy"], lw=3, ls='-', marker='v', markersize='10', label='Entropy')

plt.legend()
plt.xlabel('Victim Architecture')
plt.ylabel('Agreement(%)')
plt.tight_layout()
plt.savefig(out_file, dpi=300)
```

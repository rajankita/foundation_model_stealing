```python
import matplotlib.pyplot as plt
import csv
import pandas as pd
import seaborn as sns
# sns.set()
sns.set_context('paper')
```

```python
filename = '../results/agreements_victim_thief_lp.csv'
out_file = '../results_camera_ready/agreements_victim_thief_lp.png'

df0 = pd.read_csv(filename, index_col='Victim arch')
df0.columns = [(e) for e in df0.columns]
df0_t = df0.T
print(df0_t)
```

```python
x = df0.columns
print(x)
sns.set_style("whitegrid")
sns.set_context("paper", font_scale=2.8)

sns.set_palette("tab10", 3)
fig = plt.figure(figsize=(12,7))
ax = fig.gca()
# l1 = ax.plot(x, df0.loc["CIFAR-10 victim"], lw=4, ls='--', marker='o', markersize='12', label='CIFAR-10 victim', alpha=0.5)
# l2 = ax.plot(x, df0.loc["Indoor-67 victim"], lw=4, ls='--', marker='v', markersize='12', label='Indoor-67 victim', alpha=0.5)
# l3 = ax.plot(x, df0.loc["Caltech-256 victim"], lw=4, ls='--', marker='^', markersize='12', label='Caltech-256 victim', alpha=0.5)
l1 = ax.plot(x, df0.loc["CIFAR-10"], lw=4, ls='-', marker='o', markersize='12', label='CIFAR-10')
l2 = ax.plot(x, df0.loc["Indoor-67"], lw=4, ls='-', marker='v', markersize='12', label='Indoor-67')
l3 = ax.plot(x, df0.loc["Caltech-256"], lw=4, ls='-', marker='^', markersize='12', label='Caltech-256')

plt.legend(loc='upper left')
plt.xlabel('Victim Architecture')
plt.ylabel('Agreement(%)')
plt.xticks(rotation=30)
# plt.ylim(35,119)
plt.tight_layout()
plt.savefig(out_file, dpi=300)
```

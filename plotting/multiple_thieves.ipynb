```python
x = df0.columns
sns.set_style("whitegrid")
sns.set_context("paper", font_scale=2.8)

sns.set_palette("tab10", 5)
fig, ax = plt.subplots(1, 3, figsize=(30,11))
# ax = fig.gca()
ax[0].plot(x, df0.loc["Victim Acc"], lw=4, ls='--', marker='o', markersize='12', label='Victim Accuracy')
ax[0].plot(x, df0.loc["ResNet-34"], lw=4, ls='-', marker='v', markersize='12', label='ResNet-34')
ax[0].plot(x, df0.loc["ResNet-101"], lw=4, ls='-', marker='+', markersize='12', label='ResNet-101')
ax[0].plot(x, df0.loc["ViT-B/16"], lw=4, ls='-', marker='^', markersize='12', label='ViT-B/16')
ax[0].plot(x, df0.loc["ViT-L/16"], lw=4, ls='-', marker='>', markersize='12', label='ViT-L/16')
ax[0].plot(x, df0.loc["ViT-B/16 CLIP"], lw=4, ls='-', marker='X', markersize='12', label='ViT-B/16 CLIP')

ax[0].set_xticklabels(x, rotation=30)
ax[0].set_xlabel('Victim Architecture')
ax[0].set_ylabel('Agreement(%)')
ax[0].set_title('(a) CIFAR-10', y=-0.75, fontsize=32)


ax[1].plot(x, df1.loc["Victim Acc"], lw=4, ls='--', marker='o', markersize='12', label='Victim Accuracy')
ax[1].plot(x, df1.loc["ResNet-34"], lw=4, ls='-', marker='v', markersize='12', label='ResNet-34')
ax[1].plot(x, df1.loc["ResNet-101"], lw=4, ls='-', marker='+', markersize='12', label='ResNet-101')
ax[1].plot(x, df1.loc["ViT-B/16"], lw=4, ls='-', marker='^', markersize='12', label='ViT-B/16')
ax[1].plot(x, df1.loc["ViT-L/16"], lw=4, ls='-', marker='>', markersize='12', label='ViT-L/16')
ax[1].plot(x, df1.loc["ViT-B/16 CLIP"], lw=4, ls='-', marker='X', markersize='12', label='ViT-B/16 CLIP')

ax[1].set_xlabel('Victim Architecture')
ax[1].set_ylabel('Agreement(%)')
ax[1].tick_params(axis='x', rotation=30)
ax[1].set_title('(b) Indoor-67', y=-0.75, fontsize=32)


ax[2].plot(x, df2.loc["Victim Acc"], lw=4, ls='--', marker='o', markersize='12', label='Victim Accuracy')
ax[2].plot(x, df2.loc["ResNet-34"], lw=4, ls='-', marker='v', markersize='12', label='ResNet-34')
ax[2].plot(x, df2.loc["ResNet-101"], lw=4, ls='-', marker='+', markersize='12', label='ResNet-101')
ax[2].plot(x, df2.loc["ViT-B/16"], lw=4, ls='-', marker='^', markersize='12', label='ViT-B/16')
ax[2].plot(x, df2.loc["ViT-L/16"], lw=4, ls='-', marker='>', markersize='12', label='ViT-L/16')
ax[2].plot(x, df2.loc["ViT-B/16 CLIP"], lw=4, ls='-', marker='X', markersize='12', label='ViT-B/16 CLIP')

ax[2].tick_params(axis='x', rotation=30)
ax[2].set_xlabel('Victim Architecture')
ax[2].set_ylabel('Agreement(%)')
ax[2].set_title('(c) Caltech-256', y=-0.75, fontsize=32)


handles, labels = ax[2].get_legend_handles_labels()
lgd = fig.legend(handles, labels, loc='upper center', ncol=6, bbox_to_anchor=(0.5,1.1), fontsize=30)
plt.tight_layout()
# plt.savefig(out_file, dpi=300, bbox_extra_artists=(lgd,), bbox_inches='tight')
```
